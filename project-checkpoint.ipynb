```python
import re
import string
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, confusion_matrix, classification_report,
                              roc_curve, auc)
from sklearn.preprocessing import LabelEncoder

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
RANDOM_STATE = 42
```

```python
df = pd.read_csv("Phishing_email.csv")
df = df.drop(columns=[c for c in df.columns if "Unnamed" in c])
print("shape:",df.shape)
df.head()
```

```python
print(df['Email Type'].value_counts())
print("\nMissing values:\n",df.isnull().sum())
```

```python
fig, ax = plt.subplots()
df['Email Type'].value_counts().plot(kind='bar', color=['#2ecc71', '#e74c3c'], ax=ax)
ax.set_title("Class Distribution: Safe vs Phishing Emails")
ax.set_xlabel("Email Type")
ax.set_ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("class_distribution.png", dpi=120)
plt.show()
```

```python
df = df.dropna(subset=['Email Text']).reset_index(drop=True)
df['Email Text'] = df['Email Text'].astype(str)

def clean_text(text):
    text = re.sub(r'<.*?>', ' ', text)                     # remove HTML tags
    text = re.sub(r'http\S+|www\.\S+', ' ', text)          # remove URLs
    text = re.sub(r'\S+@\S+', ' ', text)                    # remove emails
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\d+', ' ', text)                        # remove numbers
    text = re.sub(r'\s+', ' ', text).strip()                # collapse whitespace
    return text

df['clean_text'] = df['Email Text'].apply(clean_text)
df[['Email Text', 'clean_text']].head(3)
```

```python
df = df[df['clean_text'].str.len() > 0].reset_index(drop=True)
print("Rows after cleaning:", df.shape[0])
```

```python
def count_urls(text):
    return len(re.findall(r'http\S+|www\.\S+', text))

df['has_url'] = df['Email Text'].apply(lambda t: 1 if count_urls(t) > 0 else 0)
df['num_urls'] = df['Email Text'].apply(count_urls)
df['num_exclaim'] = df['Email Text'].apply(lambda t: t.count('!'))
df['num_digits'] = df['Email Text'].apply(lambda t: sum(c.isdigit() for c in t))
df['text_length'] = df['Email Text'].apply(len)
df['num_words'] = df['clean_text'].apply(lambda t: len(t.split()))
df['uppercase_ratio'] = df['Email Text'].apply(
    lambda t: sum(1 for c in t if c.isupper()) / max(len(t), 1)
)

meta_cols = ['has_url', 'num_urls', 'num_exclaim', 'num_digits',
             'text_length', 'num_words', 'uppercase_ratio']
df[meta_cols].describe()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
compare_cols = ['has_url', 'num_urls', 'num_exclaim', 'uppercase_ratio', 'text_length', 'num_words']
for ax, col in zip(axes.flatten(), compare_cols):
    sns.boxplot(data=df, x='Email Type', y=col, ax=ax, showfliers=False)
    ax.set_title(col)
plt.tight_layout()
plt.savefig("metadata_feature_comparison.png", dpi=120)
plt.show()
```

```python
le = LabelEncoder()
df['label'] = le.fit_transform(df['Email Type'])  
print(dict(zip(le.classes_, le.transform(le.classes_))))
```

```python
X_text = df['clean_text']
X_meta = df[meta_cols].values
y = df['label'].values

X_text_train, X_text_test, X_meta_train, X_meta_test, y_train, y_test = train_test_split(
    X_text, X_meta, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

tfidf = TfidfVectorizer(max_features=5000, stop_words='english',ngram_range=(1,2),min_df=3)
X_tfidf_train = tfidf.fit_transform(X_text_train)
X_tfidf_test = tfidf.transform(X_text_test)

from scipy.sparse import hstack, csr_matrix
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_meta_train_scaled = scaler.fit_transform(X_meta_train)
X_meta_test_scaled = scaler.transform(X_meta_test)

X_train = hstack([X_tfidf_train, csr_matrix(X_meta_train_scaled)]).tocsr()
X_test = hstack([X_tfidf_test, csr_matrix(X_meta_test_scaled)]).tocsr()

print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1),
    "Naive Bayes": MultinomialNB(),
    "Neural Network (MLP)": MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=200,
                                           random_state=RANDOM_STATE, early_stopping=True),
}

results = {}
trained_models = {}

for name, model in models.items():
    print(f"Training {name} ...")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1] if hasattr(model, "predict_proba") else None

    results[name] = {
        "accuracy": accuracy_score(y_test, y_pred),
        "precision": precision_score(y_test, y_pred),
        "recall": recall_score(y_test, y_pred),
        "f1": f1_score(y_test, y_pred),
        "y_pred": y_pred,
        "y_proba": y_proba,
    }
    trained_models[name] = model
    print(f"  Done. Accuracy={results[name]['accuracy']:.4f}  F1={results[name]['f1']:.4f}")
```

```python
results_df = pd.DataFrame({
    name: {k: v for k, v in r.items() if k in ['accuracy', 'precision', 'recall', 'f1']}
    for name, r in results.items()
}).T.sort_values('f1', ascending=False)

results_df.style.format("{:.4f}").background_gradient(cmap='Greens')
```

```python
results_df[['accuracy', 'precision', 'recall', 'f1']].plot(kind='bar', figsize=(10, 6))
plt.title("Model Comparison: Accuracy, Precision, Recall, F1-score")
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=15)
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig("model_comparison.png", dpi=120)
plt.show()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 4.5))
class_names = le.inverse_transform([0, 1])
for ax, (name, r) in zip(axes, results.items()):
    cm = confusion_matrix(y_test, r['y_pred'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")
plt.tight_layout()
plt.savefig("confusion_matrices.png", dpi=120)
plt.show()
```

```python
plt.figure(figsize=(7, 6))
for name, r in results.items():
    if r['y_proba'] is not None:
        fpr, tpr, _ = roc_curve(y_test, r['y_proba'])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, label=f"{name} (AUC={roc_auc:.3f})")
plt.plot([0, 1], [0, 1], 'k--', alpha=0.4)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig("roc_curves.png", dpi=120)
plt.show()
```

```python
best_model_name = results_df.index[0]
print(f"Best model by F1-score: {best_model_name}\n")
print(classification_report(y_test, results[best_model_name]['y_pred'], target_names=class_names))
```

```python
feature_names = np.array(tfidf.get_feature_names_out().tolist() + meta_cols)

lr_model = trained_models["Logistic Regression"]
coefs = lr_model.coef_[0]
top_pos_idx = np.argsort(coefs)[-15:][::-1]   # pushes toward class 1 (Safe, per our encoding)
top_neg_idx = np.argsort(coefs)[:15]          # pushes toward class 0 (Phishing)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].barh(feature_names[top_neg_idx][::-1], coefs[top_neg_idx][::-1], color='#e74c3c')
axes[0].set_title(f"Top features indicating: {class_names[0]}")
axes[1].barh(feature_names[top_pos_idx][::-1], coefs[top_pos_idx][::-1], color='#2ecc71')
axes[1].set_title(f"Top features indicating: {class_names[1]}")
plt.tight_layout()
plt.savefig("logreg_feature_importance.png", dpi=120)
plt.show()
```

```python
rf_model = trained_models["Random Forest"]
importances = rf_model.feature_importances_
top_idx = np.argsort(importances)[-20:][::-1]

plt.figure(figsize=(9, 7))
plt.barh(feature_names[top_idx][::-1], importances[top_idx][::-1], color='#3498db')
plt.title("Top 20 Feature Importances (Random Forest)")
plt.xlabel("Importance")
plt.tight_layout()
plt.savefig("rf_feature_importance.png", dpi=120)
plt.show()
```

```python
meta_importance = pd.Series(rf_model.feature_importances_[-len(meta_cols):], index=meta_cols)
meta_importance.sort_values(ascending=False).plot(kind='bar', color='#9b59b6', figsize=(8,5))
plt.title("Random Forest Importance of Metadata Features")
plt.ylabel("Importance")
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig("metadata_importance.png", dpi=120)
plt.show()
```

```python
param_grids = {
    "Logistic Regression": (LogisticRegression(max_iter=1000, random_state=RANDOM_STATE),
                             {"C": [0.1, 1, 10]}),
    "Random Forest": (RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
                       {"n_estimators": [100, 200], "max_depth": [None, 30]}),
    "Naive Bayes": (MultinomialNB(), {"alpha": [0.1, 0.5, 1.0]}),
    "Neural Network (MLP)": (MLPClassifier(max_iter=200, random_state=RANDOM_STATE, early_stopping=True),
                              {"hidden_layer_sizes": [(64,32), (100,)]}),
}

base_model, grid = param_grids[best_model_name]
gs = GridSearchCV(base_model, grid, scoring='f1', cv=3, n_jobs=-1)
gs.fit(X_train, y_train)
print("Best params:", gs.best_params_)
tuned_pred = gs.predict(X_test)
print(f"Tuned {best_model_name} — Accuracy: {accuracy_score(y_test, tuned_pred):.4f}, F1: {f1_score(y_test, tuned_pred):.4f}")
```
